## Issue 2
### sequences of 32-bit words

```python
#Struct is added to the code to pack the 32 bit words into a byte array
import struct

#storing 32 bit words in a list
words= [10,15,20,25,30]

#Packing the words up
word_bytes = struct.pack('<5I', *words)

print(word_bytes)
```

### Explanation
Each 32 bit word takes up 4 bytes
Each bit represents the value of either 0 or 1
The position of each bit determines its value

## Issue 4
### 512-bit message blocks

```python
#Make a message 
message = "This is a 512 bit message block for testing purposes."

# Convert the message into binary
binary_message = ''.join(format(ord(char), '08b') for char in message)

# Pad the message to 512 bits
padded_message = binary_message.ljust(512, '0')

#Print the message
print("Original message:", message)

#print the binary message
print("Original message total bits:", len(binary_message))

#print the padded message
print("512-bit message block:", padded_message)

#print the total block size
print("Total block size:", len(padded_message), "bits")

```

### Explanation 
This process is done to pad the message up to the necessary level of bits required
