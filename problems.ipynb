## Issue 4
### 512-bit message blocks
### Example

```python
#Make a message 
message = "This is a 512 bit message block for testing purposes."

# Convert the message into binary
binary_message = ''.join(format(ord(char), '08b') for char in message)

# Pad the message to 512 bits
padded_message = binary_message.ljust(512, '0')

#Print the message
print("Original message:", message)

#print the binary message
print("Original message total bits:", len(binary_message))

#print the padded message
print("512-bit message block:", padded_message)

#print the total block size
print("Total block size:", len(padded_message), "bits")

```

### Explanation 
This process is done to pad the message up to the necessary level of bits required


## Issue 5
### The final 256-bit hash value
### Example

```python
#Used to create a hash value for the message
import hashlib

#Message to be hashed
message = "256 bit hash value for this message"

#Convert message to bytes
message_bytes = message.encode('utf-8')

#This will generate the hash value using SHA-256
hash_value = hashlib.sha256(message_bytes).hexdigest()

#Print the message and the hash value
print("Original message:", message)
print("SHA-256 hash:", hash_value)

#Display hash length
print("Hash length:", len(hash_value) * 4, "bits")
```

### Explanation
Secure Hash Algorithm 256-bit(SHA 256) is a encryption method which converts variables into a secure 256 bit variable
By converting the message into bytes at a 256 variable it is more secure and harder for people to access
This is done for security as it makes it a lot easier to see if data has been tampered with


## Issue 6
### Big-endian integers
### Example

```python
#Set the integer to be converted
number = 20262004

#Convert integer to 4 bytes using big-endian
big_endian = number.to_bytes(4, byteorder='big')

#Showcase the numbers 
print("Original integer:", number)
print("Big-endian bytes:", big_endian)
print("Hexadecimal:", big_endian.hex())
```

### Explanation
The  big endian integers will work by converting the bytes from largest to smallest of it

This is done mainly to make it more readable and make the data more compatible across multiple devices
